# Data Cleaning: Healthcare Dataset

This notebook handles the initial data cleaning and preprocessing for the healthcare analytics project. We will address text standardization, data type conversion, and missing values.

```python
import pandas as pd
import numpy as np

# Load the raw data
# Assuming this notebook is in the 'notebooks' folder and data is in 'data/raw'

df = pd.read_csv('../data/raw/healthcare_dataset.csv')

df.head()
```

## 1. Text Standardization
Cleaning up the formatting of string columns like `Name`.

```python
# Standardize the 'Name' column to title case
df['Name'] = df['Name'].str.title()

# Display changes
df[['Name']].head()
```

## 2. Date Type Conversion
Converting `Date of Admission` and `Discharge Date` to datetime objects.

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])

# Create a new feature: Length of Stay
df['Length of Stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days

df[['Date of Admission', 'Discharge Date', 'Length of Stay']].head()
```

## 3. Save Processed Data
Exporting the cleaned dataset so it can be used for EDA and Modeling.

```python
df.to_csv('../data/processed/cleaned_healthcare_dataset.csv', index=False)
print("Data cleaning steps complete. Ready for export.")
```
