# Exploratory Data Analysis: Healthcare Dataset

This notebook explores the cleaned healthcare dataset to uncover patterns and insights across operational, clinical, and financial metrics.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")


df = pd.read_csv('../data/processed/cleaned_healthcare_dataset.csv')

df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])
df['Length of Stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days

df.head()
```

## 1. Demographic Insights
Visualizing patient age distribution to understand the primary demographic served.

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['Age'], bins=20, kde=True, color='blue')
plt.title('Patient Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

## 2. Clinical Metrics
Analyzing the frequency and prevalence of different medical conditions.

```python
plt.figure(figsize=(10, 5))
sns.countplot(y='Medical Condition', data=df, order=df['Medical Condition'].value_counts().index, palette='pastel')
plt.title('Prevalence of Medical Conditions')
plt.xlabel('Number of Patients')
plt.ylabel('Medical Condition')
plt.show()
```

## 3. Financial Analysis
Exploring how billing amounts vary across different medical conditions.

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x='Medical Condition', y='Billing Amount', data=df, palette='Set2')
plt.title('Billing Amount Distribution by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Billing Amount ($)')
plt.show()
```

## 4. Operational Efficiency
Evaluating the Average Length of Stay based on the Admission Type (Emergency, Urgent, Elective).

```python
plt.figure(figsize=(10, 5))
sns.barplot(x='Admission Type', y='Length of Stay', data=df, palette='muted', errorbar=None)
plt.title('Average Length of Stay by Admission Type')
plt.xlabel('Admission Type')
plt.ylabel('Average Length of Stay (Days)')
plt.show()
```
